## 1. Encapsulation
The core of object-oriented programming (OOP) is the organization of the program by encapsulating related data and functions together in an object.

To "encapsulate" something means to enclose it. In programming, encapsulation means keeping data and the code that uses it in one place and hiding the details of exactly how they work together. For example, each instance of class `file` keeps track of what file on the disk it is reading or writing and where it currently is in that file. The class hides the details of how this is done so that programmers can use it without needing to know the details of how it was implemented.

To demonstrate this, we are first going to create a class representation of a rectangle. That is, we will provide the bottom-left corner (using the `Point` class) along with the width and height. 

It is interesting to compare this class to the `Square` class we defined last week.

We will also use a custom module called `rectangle_utils` to visualize our rectangles.

````python
import rectangle_utils

class Point: 
    
    """A class that represents and manipulates 2D points"""
    
    def __init__(self, x=0, y=0):
        """ 
        (number, number) -> None
        Initializes a new point at (x, y).
        """
        self.x = x
        self.y = y
    
    def print_point(self):
        """
        () -> None
        Prints the (x, y) coordinate for the point.
        """
        print("(" + str(self.x) + ", " + str(self.y) + ")")
````

First, create a Point object named `point_A`, with no arguments (i.e. use default parameters).

````python
point_A = Point()
point_A.print_point()
````

Now let's create a `Rectangle` class. We will represent a rectangle by its bottom-left corner (a `Point`), a width, and a height.

The class will have the following methods:
- `grow`: Expand the rectangle by multiplying its width and height.
- `move`: Move the rectangle by shifting its bottom-left corner.
- `area`: Return the area of the rectangle.
- `visualize`: Draw the rectangle on a coordinate plane so we can see it!

````python
class Rectangle:
    
    """Represents and manipulates rectangles."""
    
    def __init__(self, x1=0, y1=0, x2=0, y2=0):
        """ 
        (number, number, number, number) -> None
        Create a rectangle with bottom-left corner at (x1, y1)
        and upper-right corner at (x2, y2).
        """
        self.bottom_left_corner = Point(x1, y1)
        self.width = x2 - x1
        self.height = y2 - y1

    def grow(self, dw, dh):
        """
        (number, number) -> None
        Expands the rectangle by multiplying width by dw and height by dh.
        """
        self.width *= dw
        self.height *= dh
    
    def move(self, dx, dy):
        """
        (number, number) -> None
        Moves the rectangle dx units right and dy units up.
        """
        self.bottom_left_corner.x += dx
        self.bottom_left_corner.y += dy

    def area(self):
        """
        () -> number
        Returns the area of the rectangle.
        """
        return self.width * self.height
    
    def visualize(self):
        """
        () -> None
        Draws the rectangle on a coordinate plane with grid lines and labels.
        """
        rectangle_utils.visualize(self)
````

Let's create a rectangle and test it out.

````python
r = Rectangle(100, 50, 500, 300)
r.bottom_left_corner.print_point()
````

See the difference between methods and attributes below.

The origin `(0, 0)` is at the bottom-left of the plot and the axes scale automatically to fit the rectangle.

### So What About Encapsulation?

Recall the `Square` class we created last week. It also used the `Point` class but it used two instances: the lower left corner and the upper right corner. We could also use this approach to represent a `Rectangle`.

````python
class Rectangle:
    
    """A rectangle represented by 2 Points: lower left and upper right."""
    
    def __init__(self, x1=0, y1=0, x2=0, y2=0):
        """
        (number, number, number, number) -> None
        Initializes a rectangle with (x1, y1) as lower-left corner and 
        (x2, y2) as upper-right corner. All default to zeros.
        """
        self.lower_left = Point(x1, y1)
        self.upper_right = Point(x2, y2)
    
    def grow(self, dw, dh):
        """
        (number, number) -> None
        Expands the rectangle by multiplying width by dw and height by dh.
        """
        new_width = (self.upper_right.x - self.lower_left.x) * dw
        self.upper_right.x = self.lower_left.x + new_width

        new_height = (self.upper_right.y - self.lower_left.y) * dh
        self.upper_right.y = self.lower_left.y + new_height
    
    def move(self, dx, dy):
        """
        (number, number) -> None
        Moves the rectangle dx units right and dy units up.
        """
        self.lower_left.x += dx
        self.lower_left.y += dy
        self.upper_right.x += dx
        self.upper_right.y += dy

    def area(self):
        """
        () -> number
        Returns the area of the rectangle.
        """
        return ((self.upper_right.x - self.lower_left.x) *
                (self.upper_right.y - self.lower_left.y))
````

We could then implement the same methods as before but now using the different internal representation. Let's test it.

````python
r = Rectangle(10, 5, 110, 55)
r.lower_left.print_point()
r.upper_right.print_point()
````

Now let's move the rectangle.

````python
r.move(-10, 100)
r.lower_left.print_point()
r.upper_right.print_point()
````

Now let's try increasing the size of the rectangle.

````python
print(r.area())
r.grow(2, 2)
print(r.area())
````

As long as you know what the methods do, it doesn't matter what is inside! Think about this for a second: it means that you (or someone else) can write a class that does something, and you can use it without knowing how the class is implemented. This is tremendously helpful in reducing the stuff you need to keep in your head to program.

## 2. Printing Objects

It would be nice to not have to print each attribute one line at a time. It would be better if we could have a method so that every instance can produce a string representation of the object. To do that we need to format the attributes into a string. (In fact, we cheated above and already wrote such a method for the `Point` class.)

This could be converted into a method, using the special `__str__` method as follows.

````python
class Point:
    
    """A class that represents and manipulates 2D points"""
    
    def __init__(self, x=0, y=0):
        """ 
        (number, number) -> None
        Initializes a new point at (x, y).
        """
        self.x = x
        self.y = y
    
    def print_point(self):
        """
        () -> None
        Prints the (x, y) coordinate for the point.
        """
        print('(' + str(self.x) + ', ' + str(self.y) + ')')
````

````python
my_point = Point()
print(my_point)
my_point.print_point()
````

Notice how `print(my_point)` doesn't give us useful output? We can fix this by adding a `__str__` method to our class. This is a special method that Python calls whenever it needs to convert an object to a string (e.g., when you call `print()`).

````python
class Point:
    
    """A 2D Point, at coordinates x, y"""

    def __init__(self, x=0, y=0):
        """ 
        (number, number) -> None
        Create a new point at x, y.
        """
        self.x = x
        self.y = y
        
    def __str__(self):
        """
        () -> str
        Return the coordinates of the Point.
        """
        return "(" + str(self.x) + ", " + str(self.y) + ")"
````

````python
my_point = Point()
print(my_point)
````

````python
class Point:
    
    """A 2D Point, at coordinates x, y"""

    def __init__(self, x=0, y=0):
        """ 
        (number, number) -> None
        Create a new point at x, y.
        """
        self.x = x
        self.y = y
        
    def __str__(self):
        """
        () -> str
        Return the coordinates of the Point.
        """
        return "(" + str(self.x) + ", " + str(self.y) + ")"

class Rectangle:
    
    """Represents and manipulates rectangles."""
    
    def __init__(self, corner, w, h):
        """ 
        (Point, number, number) -> None
        Create a rectangle with bottom-left corner at corner,
        width of w, and height of h.
        """
        self.bottom_left_corner = corner
        self.width = w
        self.height = h

    def __str__(self):
        """
        () -> str
        Returns a string representation of the rectangle.
        """
        return "Bottom Left Point: " + str(self.bottom_left_corner) + "\nWidth x Height: " + str(self.width) + " x " + \
            str(self.height) + "\nArea: " + str(self.area())

    def grow(self, dw, dh):
        """
        (number, number) -> None
        Expands the rectangle by multiplying width by dw and height by dh.
        """
        self.width *= dw
        self.height *= dh
    
    def move(self, dx, dy):
        """
        (number, number) -> None
        Moves the rectangle dx units right and dy units up.
        """
        self.bottom_left_corner.x += dx
        self.bottom_left_corner.y += dy

    def area(self):
        """
        () -> number
        Returns the area of the rectangle.
        """
        return self.width * self.height
````

## 3. Building a Cash Register

### Creating a cash register type

We're going to explore classes by writing a cash register class. In the example, we will use Canadian currency including the Loonie (`$1`) and the Toonie (`$2`). We'll pass in the number of loonies, toonies, fives, tens, and twenty dollar bills.

What functionality do we want the `Cash_Register` to have?
- Print the contents (`__str__`).
- Calculate the value of the contents (`get_total`).
- Add some cash (`add`).
- Remove some cash (`remove`).

Below is the class with the constructor completed and the method signatures and docstrings for the remaining methods. Let's write the body of each method.

````python
class Cash_Register:
    """A cash register."""

    def __init__(self, loonies=0, toonies=0, fives=0, tens=0, twenties=0):
        """ 
        (int, int, int, int, int) -> None
        Creates a Cash_Register with the given number of each denomination.
        Attributes: loonies, toonies, fives, tens, twenties.
        
        >>> register = Cash_Register(4, 2, 4, 5, 5)
        >>> register.loonies
        4
        """
        self.loonies = loonies
        self.toonies = toonies
        self.fives = fives
        self.tens = tens
        self.twenties = twenties

    def __str__(self):
        """
        () -> str
        Return a string listing the count of each denomination,
        from twenties down to loonies.
        
        >>> print(Cash_Register(1, 2, 3, 4, 5))
        The cash register holds: 
        20s: 5
        10s: 4
        5s: 3
        2s: 2
        1s: 1
        """
        return "The cash register holds: \n20s: " + str(self.twenties) \
            + "\n10s: " + str(self.tens) + "\n5s: " + str(self.fives) \
            + "\n2s: " + str(self.toonies) + "\n1s: " + str(self.loonies)
        
    def get_total(self):
        """
        () -> number
        Return the total dollar value of all cash in the register.
        
        >>> Cash_Register(4, 2, 4, 5, 5).get_total()
        178
        """
        return 20 * self.twenties + 10 * self.tens + 5 * self.fives + 2 * self.toonies + self.loonies

    def add(self, count, denomination):
        """
        (int, str) -> None
        Add count items of the given denomination to the register.
        denomination is one of: 'loonies', 'toonies', 'fives', 'tens', 'twenties'.
        
        >>> register = Cash_Register()
        >>> register.add(3, 'fives')  # adds 3 five-dollar bills
        >>> register.fives
        3
        """
        if denomination == 'loonies':
            self.loonies += count
        elif denomination == 'toonies':
            self.toonies += count
        elif denomination == 'fives':
            self.fives += count
        elif denomination == 'tens':
            self.tens += count
        elif denomination == 'twenties':
            self.twenties += count    

    def remove(self, count, denomination):
        """
        (int, str) -> None
        Remove count items of the given denomination from the register.
        denomination is one of: 'loonies', 'toonies', 'fives', 'tens', 'twenties'.
        Hint: think about how you can reuse the add method.
        
        >>> register = Cash_Register(0, 0, 5, 0, 0)
        >>> register.remove(2, 'fives')  # removes 2 five-dollar bills
        >>> register.fives
        3
        """
        self.add(-count, denomination)
````

### The `Planet` Class

Each planet has a name, a distance from the sun (in pixels), a drawing radius, a color, and a starting angle on its orbit. The class should also have a `__str__` method and an `orbital_period` method that calculates how long it takes the planet to orbit the sun using Kepler's Third Law:

$$T = T_{earth} \cdot \left(\frac{distance}{distance_{earth}}\right)^{1.5}$$

where $T_{earth} = 365.25$ days and $distance_{earth} = 90$.

````python
class Planet:
    """Represents a planet in a solar system."""
    
    def __init__(self, name, distance, radius, color, angle=0):
        """
        (str, number, number, str, number) -> None
        Create a planet with a name, distance from the sun, radius,
        color, and starting angle in degrees.
        Attributes: name, distance, radius, color, angle.
        
        >>> earth = Planet('Earth', 90, 5, '#4B7BE5', 45)
        >>> earth.name
        'Earth'
        >>> earth.distance
        90
        """
        self.name = name
        self.distance = distance
        self.radius = radius
        self.color = color
        self.angle = angle
    
    def __str__(self):
        """
        () -> str
        Return the planet's name, distance, and radius as a string.
        
        >>> print(Planet('Earth', 90, 5, '#4B7BE5'))
        Earth (distance: 90, radius: 5)
        """
        return self.name + " (distance: " + str(self.distance) + ", radius: " + str(self.radius) + ")"
    
    def orbital_period(self):
        """
        () -> float
        Calculate the orbital period in days using Kepler's Third Law:
            T = T_earth * (distance / distance_earth) ^ 1.5
        where T_earth = 365.25 days and distance_earth = 90.
        
        >>> Planet('Earth', 90, 5, '#4B7BE5').orbital_period()
        365.25
        """
        earth_distance = 90
        earth_period = 365.25
        period = earth_period * (self.distance / earth_distance) ** 1.5
        return period
````

### The `SolarSystem` Class

The `SolarSystem` class manages a collection of `Planet` objects. It should be able to add and remove planets, report the total number of planets, print a summary of the system, and visualize or animate the solar system using `solar_system_utils`.

````python
class SolarSystem:
    """Represents a solar system with planets orbiting a sun."""
    
    def __init__(self):
        """
        () -> None
        Create an empty solar system with no planets.
        Attributes: planets (list).
        
        >>> ss = SolarSystem()
        >>> ss.planets
        []
        """
        self.planets = []
    
    def __str__(self):
        """
        () -> str
        Return a string listing the number of planets and each planet's info.
        
        >>> ss = SolarSystem()
        >>> ss.add_planet(Planet('Earth', 90, 5, '#4B7BE5'))
        >>> print(ss)
        Solar System with 1 planets:
          - Earth (distance: 90, radius: 5)
        """
        result = "Solar System with " + str(self.total_planets()) + " planets:"
        for planet in self.planets:
            result += "\n  - " + str(planet)
        return result
    
    def total_planets(self):
        """
        () -> int
        Return the number of planets in the solar system.
        
        >>> ss = SolarSystem()
        >>> ss.total_planets()
        0
        """
        return len(self.planets)
    
    def add_planet(self, planet):
        """
        (Planet) -> None
        Add a planet to the solar system.
        
        >>> ss = SolarSystem()
        >>> ss.add_planet(Planet('Mars', 115, 4, '#E04040'))
        >>> ss.total_planets()
        1
        """
        self.planets.append(planet)
    
    def remove_planet(self, name):
        """
        (str) -> None
        Remove the planet with the given name from the solar system.
        
        >>> ss = SolarSystem()
        >>> ss.add_planet(Planet('Mars', 115, 4, '#E04040'))
        >>> ss.remove_planet('Mars')
        >>> ss.total_planets()
        0
        """
        for planet in self.planets:
            if planet.name == name:
                self.planets.remove(planet)
    
    def visualize(self):
        """
        () -> None
        Draw a static visualization of the solar system.
        """
        solar_system_utils.visualize(self)
    
    def animate(self, frames=150, dt=0.05):
        """
        (int, float) -> None
        Animate the solar system with planets orbiting the sun.
        """
        solar_system_utils.animate(self, frames, dt)
````
