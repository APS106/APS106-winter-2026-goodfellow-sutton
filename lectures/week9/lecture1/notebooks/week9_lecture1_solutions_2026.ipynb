## 1. Difficult Dictionaries
This section contains the solutions for this section focused on Dictionaries.

```python
# Q1 What will this print? - Dictionary Indexing

d = {"Running":45, "Hiking":101, "Biking":29, "Walking":9000}

d['Running'] += 1
print(d['Running'])

# SOLUTION: 46
# we add one to the value at 'Running' and store it back in. So the key/value "Running":45 becomes "Running":46.
```

```python
# Q2 What will this print? - Dictionary del/pop

d = {"Running":45, "Hiking":101, "Biking":29, "Walking":9000}

del d['Walking']
print(d.pop('Running'))
print(list(d.values()))

# SOLUTION: 
# 45
# [101, 29]

# del will delete the key/value pair with key 'Walking'
# pop will return the value of 'Running' and remove it (pop it!) from the dictionary
# Both 'Walking' and 'Running' are therefore removed from the dictionary, so neither show up in our values, only Hiking/Biking remain
# the 'in' here returns a boolean
```

```python
# Q3 - What will this print? - Dictionary nested/'in'

dict_1 = {
    "class": {
        "student": {
            "name": "Mike",
            "marks": {
                "physics": {
                    "midterm": 67, 
                    "exam": 81 }
            }
        }
    }
}

print("class" in dict_1)
print("name" in dict_1["class"])
print(dict_1["class"]["student"]["name"]["marks"]["physics"])

# SOLUTION: 
# True
# False
# ERROR

# class is directly in dict_1 as a key
# name is NOT a key in dict_1["class"]. It's a key for the sub-dictionary dict_1["class"]["student"]
# this errors - "name" and "marks" are keys at the same level of the dictionary. If we take the value of key "name", there is no value that is a dictionary with the key "marks". So it errors
```

```python
# Q4 What will this print? - Dictionary + Split

string = "so APS106 is so cool, isn't it?"
str_split = string.split(' ')
dictionary = {}

for s in (str_split):
    dictionary[s] = 1
print(dictionary)

# SOLUTION: it prints
# {'so': 1, 'APS106': 1, 'is': 1, 'cool,': 1, "isn't": 1, 'it?': 1}
# split will break it into a list with each word, split by the spaces
# Then, we add each one to a dictionary. 
# We won't make a duplicate entry for 'so' even though it shows up twice, since the key used for the dictionary is the same.
# Note that the order doesn't matter.
```

```python
# Q1 What will this print? - split + lists
string = "APS 106 is really cool!"
string_split = string.split(' ')

print(string_split[0] + string_split[1] + ' ' + string_split[4])

print(string_split)

# SOLUTION: APS106 cool!
# Split gives us a list of strings split up by our space
# so, .split(' ') gives us string_split = ['APS', '106', 'is', 'really', 'cool!']
# Then we stitch it together with + and spaces
```

```python
# Q2 What will this print? - List comparisons + Aliasing

colours = ['pink', 'blue', 'orange']
colours2 = colours
print(colours.extend(["Red", "yellow"]))
print(max(colours))
print(min(colours2))

# SOLUTION: 
#None
#yellow
#Red
# we assign colours2 the value of colours. With list aliasing, colours2 and colours now point at the same list, so any modifications to either will affect both
# append returns 'None' as it's return type
# our max and min are doing string comparisons! The maximum one is 'yellow' as it starts with a 'y' which is the highest ASCII representation starting letter from the words here.
# the min is 'Red', as 'R' is the lowest ASCII representation starting letter out of the words we have here.
```

```python
# Q3 What will this print? - List append vs. extend

colours = ['red']
colours.append('pink')
colours.append(['orange'])
colours.extend(['blue','yellow'])
colours.extend('purple')
print(colours)

# Options:
A. ['red', 'pink', ['orange'], 'blue', 'yellow', 'p', 'u', 'r', 'p', 'l', 'e']
B. ['red', 'pink', ['orange'], 'blue', 'yellow', 'purple']
C. ['red', 'pink', 'orange', ['blue', 'yellow'], 'purple']
D. ['red', 'pink', ['orange'], ['blue', 'yellow'], 'purple']
E. ['red', 'pink', 'orange', 'blue', 'yellow', 'purple'']

# SOLUTION: 
# ['red', 'pink', ['orange'], 'blue', 'yellow', 'p', 'u', 'r', 'p', 'l', 'e']
# append will add one element to the end
# extend will add each element from an iterable to the end of the list, one by one
# here, we append 'pink' to add it to the end
# then we append the list ['orange'] to add it to the end
#, extend ['blue','yellow'] to add them both, 
# then extend 'purple', which adds each letter of purple one by one
```

```python
# Q4 What will this print?

super_list = [
    ["Is", "APS", "106", "super lame", "?"],
    ["APS", "112", "is", "the most fun I've ever had", "yes definitely!"]
]

print(f"How do I feel about APS 106? {super_list[0][3].split(' ')[0] + super_list[1][-1][-1]}")

# SOLUTION: How do I feel about APS 106? super!
# this is just tricky indexing. We the first index is selecting which row, with the next index selecting which item in the row to look at. Then split is used to split "super lame" into two parts and we take the first [0].
# Just subbing it in and evaluating one index at a time:
# super_list[0][3].split(' ')[0] -> ["Is", "APS", "106", "super lame", "?"][3].split(' ')[0] ->  "super lame".split(' ')[0] -> ["super", "lame"][0] -> "super"
# super_list[1][-1][-1] -> ["APS", "112", "is", "the most fun I've ever had", "yes definitely!"][-1][-1] -> "yes definitely!"[-1] -> "!"
# adding it all together, we get "super!" added onto the print statement.
```

```python
# Q2 What will this print? - Tuples!

tupl = (5, 2, 72)
x, y, z = tupl

print(x ** y + z + len(tupl))

# SOLUTION: 100
# We unpack our tuple and store the values into x, y, z
# Then, we do a math operation on it. 5**2 + 72 = 97. Then we add len(tupl). It's a tuple length 3, so this returns 3. 97 + 3 = 100
```

```python
#Q3 - What will this print? Tuples

def power_up_info():
    power_ups = ('Mushroom', 'Fire Flower', 'Super Star', '1-Up', 'Mega Mushroom')
    return power_ups, len(power_ups)

new_power_ups, count = power_up_info()
a, b, c, d, e = new_power_ups
cde = [c, d, e]

print(a, count, cde[::-2])

#A) ValueError: too many values to unpack
#B) Mushroom 5 ('Mega Mushroom', 'Super Star')
#C) TypeError: 'tuple' object does not support item assignment
D) Mushroom 5 ['Mega Mushroom', 'Super Star']

# SOLUTION: D
# We get our list of power ups from the function, returning:
# ('Mushroom', 'Fire Flower', 'Super Star', '1-Up', 'Mega Mushroom'), 5
# then, we unpack them and cde takes [c, d, e][::-2], so taking every 2nd element going backwards. 
# So, cde[::-2] will  take ['Mega Mushroom', 'Super Star']
# note that it's square brackets and not round as this is a list that we're indexing rather than a tuple.
```

```python
#Q4 - What will this print? Tuple methods

warp_zones = (
    ('Underground', ['Level 1', 'Level 2'], 100),
    ('Sky', ['Level 2', 'Level 3', 'Final Level'], 200),
    ('Castle', ['Level 3', 'Level 4'], 300),
    ('Underwater', ['Level 3', 'Level 4', 'Final Level'], 400),
)

x = warp_zones[-1][1][-1]
for warp in warp_zones:
    if x in warp[1]:
        print(warp[0])

# SOLUTION:
# Sky
# Underwater

# warp_zones is a tuple of length 4, with tuples inside. The inner tuples seem to be
# (Name, List[Levels], Some Int).
# Our code grabs x = warp_zones[-1][1][-1], which means we index the last inner tuple [-1], then the index 1 element [1], then the last element of this list [-1]
# This gets us ('Underwater', ['Level 3', 'Level 4', 'Final Level'], 400)[1][-1] -> ['Level 3', 'Level 4', 'Final Level'][-1] -> 'Final Level'
# Then, we search through all tuples and if their list contains final level, we print their name
# the ones that have 'Final Level' are 'Sky' and 'Underwater'
```

## 4. For a While There
This section contains the solutions for this section focused on For loops.

```python
# Q1 - What indices are 'B' and 'C' at?

ls = [
    ['A', 'A', 'A', 'A'],
    ['A', 'A', 'A', 'E', 'B'],
    ['A', 'A', 'A'],
    ['A', 'C', 'A', 'D']
]

print(ls[1][4]) #B
print(ls[3][1]) #C

# SOLUTION:
# ls[1][4] == "B"
# ls[3][1] == "C"
```

```python
#Q2 - What will this print? Range

test_string = "A3DE4"
for i in test_string:
    if i.isdigit():
        print(int(i)**2)

# SOLUTION:
# 9
# 16

# we iterate through test string
# for each character, we check if it's a digit. If it's a digit, we print that digit's squared value.
# The only digits are 3 and 4. These give 3**2 = 9 and 4**2 = 16
```

```python
# Q3 What is the output of the code?

# Nested lists
ls = [
    [1, 1, 1, 1, 2],
    [1, 1, 1, 1, 2],
    [1, 16, 1, 1, 2],
    [2, 1, 1, 1, 4],
]
counts = []
ncol = len(ls[0])
nrow = len(ls)

for col in range(ncol):
    c = 0
    for row in range(nrow):
        c += ls[row][col]
    counts.append(c)
print(counts)

# SOLUTION: [5, 19, 4, 4, 10]
# This code prints the sum of all columns. We get the number of columns and number of rows using len, then we loop through the columns one by one, adding the row value.
```

```python
# Q4 - What is the output of the code?
my_food = ['apples', 'potato', 'pears', 'pomegranate']

dictionary = {}
for food in my_food:
    count = 0 
    for letter in food:
        if letter >= 'a' and letter < 'f':
            count += 1
    dictionary[food] = count

print(dictionary)
    
# SOLUTION: 
# {'apples': 2, 'potato': 1, 'pears': 2, 'pomegranate': 4}

# We're iterating through all food items in my_food
# for each of these, we iterate through all letters in the food item, as strings are iterable
# we're counting all letters >= a and < f (therefore, any letters a, b, c, d, e)
# finally, we create a new entry in our dictionary for each count we found.
# apples has 2 - a, e. Potato has 1 - a. Pears has 2 - e, a. Pomegranate has 4 - e, a, a, e
```

## 5. Final Jeopardy
This section contains the solutions for the Final Jeopardy section.

Write a function `filter_and_organize_books` that:
- Takes a **list of book tuples** (`book_list`) and a float (`min_price`).
    - Each tuple contains: =`(title (str), author (str), price (float), genre (str))`.
- **Keeps only books** with `price >= min_price`.
- Organizes them into a dictionary `{genre: [book titles]}`, where:
    - **Keys** are genres.
    - **Values** are lists of book titles.
    - A **genre is added only if at least one book meets the price condition.**
- **Returns** the dictionary. The function **must not print anything**.   

Example input:
`books = [
("The Hobbit", "J.R.R. Tolkien", 12.99, "Fantasy"),
("A Brief History of Time", "Stephen Hawking", 15.50, "Science"),
("The Great Gatsby", "F. Scott Fitzgerald", 10.99, "Classic"),
("1984", "George Orwell", 9.99, "Dystopian"),
("The Catcher in the Rye", "J.D. Salinger", 8.99, "Classic"),
("The Theory of Everything", "Stephen Hawking", 14.75, "Science"),
("Brave New World", "Aldous Huxley", 11.50, "Dystopian"),
]`

Expected output:
`{
'Fantasy':['The Hobbit'],
'Science': ['A Brief History of Time', 'The Theory of Everything'],
'Classic': ['The Great Gatsby'],
'Dystopian': ['Brave New World']
}`

```python
def filter_and_organize_books(book_list, min_price):
    '''
    List(Tuple(str, str, float, str)) -> Dictionary(str:list(str))
    '''
    dictionary = {}

    # iterate through all book tuples in our list
    for book_tuple in book_list:
        # note that our tuple is (title, author, price, genre)
        # you can decompose this or keep it as a tuple and index say book_tuple[0] as you go to get individual elements
        # I'll break it down and assign to variables (title, author, price, genre) for ease of reading
        (title, author, price, genre) = book_tuple

        # check, is the price (index 2) greater than the minimum?
        if price > min_price:
            # if so, include
            # check if genre is already in dictionary
            if genre in dictionary.keys():
                # if so, just append title to our list
                dictionary[genre].append(title)
            else:
                # if not, create a new list with only our new title
                dictionary[genre] = [title]

    # Return!
    return dictionary

books = [
("The Hobbit", "J.R.R. Tolkien", 12.99, "Fantasy"),
("A Brief History of Time", "Stephen Hawking", 15.50, "Science"),
("The Great Gatsby", "F. Scott Fitzgerald", 10.99, "Classic"),
("1984", "George Orwell", 9.99, "Dystopian"),
("The Catcher in the Rye", "J.D. Salinger", 8.99, "Classic"),
("The Theory of Everything", "Stephen Hawking", 14.75, "Science"),
("Brave New World", "Aldous Huxley", 11.50, "Dystopian"),
]

print(filter_and_organize_books(books, 10))
```
