```python
# Q2 What will this print? - Deceptive Dictionaries

count = {}
count[(1, 2, 4)] = 5
count[(4, 2, 1)] = 7
count[(1, 2)] = 6
count[(4, 2, 1)] = 2
tot = 0

for i in count:
    tot += count[i]

print(len(count) + tot)

# a) 25
# b) 17
# c) 16
# d) TypeError occurs as tuples can't be made keys of a dictionary

# SOLUTION: 
# c) 16
# Tuples can be made keys of a dictionary as they're immutable! So d) is not the answer.
# We have 3 keys: (1,2,4), (4,2,1), and (1,2). The values of these after the code runs are 5, 7, 6, then the value at key (4,2,1) gets overwritten to 2, giving values of 5, 2, 6.
# our loop goes through these three keys and sums the values to 5 + 2 + 6 = 13. 
# As we have 3 entries (key:value pairs) in our dictionary, len(count) = 3
# we print len(count) + tot = 3 + 13 = 16
```

```python
# Q3 - What will this print? - Alias

def f1(x):
    return x

def f2(x):
    return list(x)

x = [1, 2, 3]

print(x is f1(x))
print(x is f2(x))
print(x == f2(x))

y = x
x.append(24)
print(x == y)

# SOLUTION: 
# True
# False
# True
# True

# This question is about aliasing
# is checks if the two objects are the exact same object (ID) in memory, while == checks if they have the same values
# f1(x) returns the object x itself, which is the same as x - True
# f2(x) returns a list(x), which creates a new list - False
# However, the values in f2(x)'s return are the same as the original x, so == returns True - True
# We set a new var y = x, which is an alias. When we append 24 to x, it also adds to where y is pointing (the same place)
# so, x == y and x is y are both true
```

```python
# Q4 What will this print? - Dubious Dictionaries and Suspicious Sets!

phonebook = {'Akako' : 'Ke',
             'Ke': 'Freya',
             'Alya': 'Freya',
            'Freya': 'Frann'}

birthdays = {'September': ['Ke', 'Freya'],
             'March': ['Alya', 'Frann'],
            'July': ['Akako']}

names = set(phonebook.keys())
friends = set(phonebook.values())
check = names.intersection(friends)
print(check)

bday1 = set(birthdays['September'][1])
bday2 = set(birthdays['March'][1])
check = bday1.intersection(bday2)
print(check)

# SOLUTION: 
# {'Freya', 'Ke'}
# {'F', 'r', 'a'}
# We setup our phonebook
# We extract our keys as a set for names and the values as a set for values
# We get 'check' as an intersection of these two sets, so all values that are in both - this gives Freya and Ke

# Then, we setup sets with the items from birthdays, which is a dictionary of string:list (key:value). We create a set from birthdays['September'][1] which is 'Freya'. So our set has all unique letters {'F', 'r', 'e', 'y', 'a'}
# We do the same with Frann
# Then, we take an intersection which gives all letters common to the two, giving {'F', 'r', 'a'}
```

```python
# Q5 - What will this print? - Devilish Dictionaries

trade_inventory = {
    'Emerald': 5,
    'Gold': 10,
    'Steel': 3,
    'Diamond': 8
}

x = {}
for a, b in trade_inventory.items():
    x[b] = a

print(x.pop(min(x)))
x[max(x)] = "Silver"
print(x)

# SOLUTION:
# Steel
# {5: 'Emerald', 10: 'Silver', 8: 'Diamond'}

# we initialize our dictionary trade_inventory
# our look goes through all items as key,value pairs a, b.
# and we set the keys as b and values as a (flipping the order)
# We pop the minimum value key which is 3: 'Steel'. So the value 'Steel' gets printed and removed (with it's key) from the dictionary by pop().
# Then, we index at the max value key (10) and replace the value 'Gold' with 'Silver'
```

## 2. Read, Write, maybe even Append
This section contains the solutions for this section focused on Files + CSVs.

```python
# Q1 What will be stored in myfile? - write

names = ['Charlie', 'Seb', 'Chris']
names.sort()
myfile = open("names.txt", "w")
for item in names:
    myfile.write(item + '\n')
myfile.close()
    
# SOLUTION:
# Charlie
# Chris
# Seb
# these will all be on different lines!

# We sort the list, so Chris gets put before Seb (alphabetical sorting)
# Then, we open a file and write to it, writing each name separated by a newline.
```

```python
# Q2 What will this print? - readlines()

f = open("grades.txt", "r")
for line in f.readlines():
    print(line.strip())
f.close()

# SOLUTION:
# John B-
# Susmit A-
# Eva C+
# Seb F
# Chris C+
# We open grades.txt. For each line, we print the line stripped.
# Note that if we didn't have .strip(), an extra newline would show up, as we'd print the line literal "John B-\n", then print adds another newline at the end.
```

```python
# Q3 What will this print? - csvReader

import csv

with open('test.csv', newline = '') as csvfile:
    filereader = csv.reader(csvfile)
    for row in filereader:
        print(row[0] + ' exam on ' + row[2] + 'th at ' + row[-1])

# SOLUTION: 
# APS111 exam on 20th at 9:30
# APS111 exam on 20th at 2:00
# APS164 exam on 18th at 9:30

# We open test.csv as csvfile
# Then, we read the csv using .reader.
# we loop through each row and extract info from the row, with each element separated by a comma being an element in a list
# row[0] gives the class name (0th element), row[2] gives the date (3rd column, index 2), and row[-1] gives the last element which is the time.
```

```python
# Q4 What will this print? - for line

counts = {}

with open('animals.txt', 'r') as animals:
    for line in animals:
        if "cat" in line:
            if "cat" not in counts:
                counts["cat"] = 3
            else:
                counts["cat"] += 1

print(counts)
# SOLUTION:
# {'cat': 5}

# with open lets us open the text file and assign to animals, automatically closing at the end of the with block
# we go through each line. If the line contains the substring "cat", we go into the IF block
# the IF block checks if we've already added "cat" as a key to counts. If not we init it to 3. Otherwise, we add 1 to this counter.
# As "cat" shows up in 3 lines, we get 3 + 1 + 1 = 5
```

```python
# Q5 - What will this print? - readline()

dict1 = {}
with open("money.txt") as money_moves:
    line = money_moves.readline()
    while line != '':
        split = line.strip().split(',')
        dict1[split[0]] = split[1]
        line = money_moves.readline()

print(dict1)

# SOLUTION:
# {'loonies': '1.00', 'toonies': '2.00', 'pennies': '0.01', 'dime': '0.10'}

# We init an empty dictionary dict1
# with open lets us open the text file and assign to money_moves, automatically closing at the end of the with block
# we use the 'readline()' approach, which means we read one line at a time. Each time we call readline(), we advance one line down the file, ending when we hit an empty string ''
# For each line, we strip (remove spaces and newlines), and split according to commas
# So, each line is split into the first item and the second item cleaned up.
# Then, we create entries in the dictionary with the key being the first item ('loonies', 'toonies', etc...) 
# and the second being the value '1.00' or '2.00' or etc...
```

```python
# Q2 - What will this print?

class Journey:
    def __init__(self, origin, destination, duration):
        self.origin = origin
        self.destination = destination
        self.duration = duration

x = Journey('X', 'Y', 2.5)

print(isinstance(x, Journey))
print(isinstance(x, object))
print(isinstance(x.origin, str))

# SOLUTION:
# True
# True
# True

# This question is about isinstance, which checks if the type of the first argument is the second argument (e.g., is x a Journey?)
# x is a Journey - True
# x is an object - True (everything is an object in python)
# x.origin is a string - True
```

```python
#Q3 - What will this print?
class Animal:
    def __init__(self, name):
        self.weight = 16
        self.name = name
        self.age = 10

    def __str__(self):
        return self.name + " meows"

    def speak(self):
        print(self.name + " says hi", end = '')
        return 'skers'

cat = Animal("Whiskers")
print(cat.speak())
print(cat)

# SOLUTION:
# Whiskers says hiskers
# Whiskers meows
# We initialize an animal with the self.name = "Whiskers"
# We call the cat.speak() method which prints (self.name + " says hi", end = ''), giving "Whiskers says hi" with no ending newline
# this method returns 'skers' which we print out from the print(cat.speak()) line. This gets added to the end of the previous line (as end=''), giving "Whiskers says hiskers"
# Then, we print(cat), which runs the  __str__ default method, giving self.name + " meows" or Whiskers meows
```

```python
# Q4 - What's the output?

class StudentRecord:
    def __init__(self, name, midterm=0, project=0, final=0):
        self.name = name
        self.grades = [midterm, project, final]

    def remark(self, grade_component, number):
        self.grades[grade_component] += number

grade = 90
sr = StudentRecord("Ana Rao", grade, grade + 6, final = 90)
sr.remark(1, 3)
print(sum(sr.grades)/3)

# SOLUTION:
# 93.0
# We create a new StudentRecord with name = "Ana Rao", midterm = grade = 90, project = 96, final = 90
# Then, we apply remark(1,3) to this StudentRecord. This edits grade component index 1 (the project) and adds 3 to it.
# we print the sum of all the grades/3, which is midterm = 90, project = 96+3 = 99, final = 90. The average is 90 + 99 + 90/3 = 93.0
```

```python
#Q5 - What will this print?
class CityDweller:
    def __init__(self, name, gender, city):
        self.name = name
        self.gender = gender.lower()
        self.city = city.lower()

class CityCounter:
    def __init__(self, people):
        self.people = people

    def count_city(self, city_name):
        count = 0
        for person in self.people:
            if person.city == city_name.lower():
                count += 1
        return count

# Example data and usage
people = [
    CityDweller("Alice", "Female", "London"),
    CityDweller("Bob", "Male", "Paris"),
    CityDweller("Charlie", "Male", "London")
]

counter = CityCounter(people)
print("People from London:", counter.count_city("London"))

# SOLUTION: 
# People from London: 2
# We first create a list of 3 CityDwellers. CityDwellers are people with names, genders, and city (in which they dwell)
# We then add all these (in a list) to a new class called CityCounter, which is initialised with a list of people
# We call counter.count_city("London"), passing in the city_name = "London"
# We iterate through all people in the city and check if their city is equal to "london" (lower case). If so, we add one count, to count all dwellers in London
```

## 4. Final Jeopardy
This section contains the solutions for the Final Jeopardy section.

Write a class `LibraryEmployee` that has the following attributes:
- **name**: the name of the employee.
- **books**: a list of this employee's favourite `Books`.
- **num_coffees**: an integer representing how many coffees this employee has had today. This value should begin at zero when a `LibraryEmployee` is initialized.

The `LibraryEmployee` class should have an initializer that accepts the employee's **name** and **books** as arguments. These values should be assigned to the object's respective attributes.

Additionally, the `LibraryEmployee` class should have the following methods:
- **drink_coffee()**: takes in an integer **num**, and increases this `LibraryEmployee`'s **num_coffees** attribute by that integer.
- **shush_hooligans()**: takes in an integer **num_hooligans** and prints out "Shhh!" that many times. Also drinks a coffee once each time they shush a hooligan
- A **default method** that converts the `LibraryEmployee` into a string. It should return a string of all the employee's attributes formatted nicely for viewing (don't worry too much about parsing and printing the details from the `Books`).

**BONUS**:
Also, write a class `Books` that has an initializer that sets the following attributes:
- **name**: the name of the book.
- **author**: the author of the book.
- **read_count**: the count of times this book has been read. This should be initialized to 0.

It should have a method **read()** that increments the **read_count** of that book and prints out something about the book's name.

```python
class LibraryEmployee:
    def __init__(self, name, books):
        self.name = name
        self.books = books
        self.num_coffees = 0

    def drink_coffee(self, num):
        self.num_coffees += num

    def shush_hooligans(self, num_hooligans):
        for i in range(num_hooligans):
            print("Shhh!")
            self.drink_coffee(1)

    def __str__(self):
        return f"Name: {self.name}\nFavourite Books: {self.books}\nNumber of Coffees: {self.num_coffees}"

class Book:
    def __init__(self, name, author):
        self.name = name
        self.author = author
        self.read_count = 0

    def read(self):
        self.read_count += 1
        print(f"{self.name} was a great book!")

TVHC = Book("The Very Hungry Caterpillar", "Eric Carle")
APS106 = Book("APS106 Textbook", "Sebastian Goodfellow")
HTBAH101 = Book("How to be a hooligan 101", "Chris Sutton")

charlie = LibraryEmployee("Charlie", [TVHC, APS106, HTBAH101])
charlie.drink_coffee(1)
charlie.shush_hooligans(3)
print(charlie)

charlie.books[0].read()
```
